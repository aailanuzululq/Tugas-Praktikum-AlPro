```python
%%writefile saving_project/interest.py
def compound_interest(principal, rate, time):
    return principal * (1 + rate) ** time


def simple_interest(principal, rate, time):
    return principal * (1 + rate * time)
```

```python
%%writefile saving_project/display.py


def print_result(principal, rate, years, ci, si):
    print()
    print('===== Hasil Perhitungan Tabungan =====')
    print(f'{'Saldo Awal':<18}: \u00a5{principal:,}')
    print(f'{'Bunga Tahunan':<18}: {rate * 100, 2)}%')
    print(f"{'Lama Menabung':<18}: {years} tahun")
    print('-' * 38)
    print(f'Nilai Akhir (Bunga Majemuk)  : \u00a5{ci:,.2f}')
    print(f'Nilai Akhir (Bunga Sederhana): \u00a5{si:,.2f}')
    print('=' * 38)
```

```python
%%writefile saving_project/main.py
import sys
import interest
import display


def main():
    print('=== Program Kalkulator Bunga Tabungan ===')
    principal = float(input('Masukkan saldo awal (contoh: 80000): '))
    rate = float(input('Masukkan bunga tahunan (misal 5% \u2192 0.05): '))
    years = int(input('Masukkan lama menabung (tahun): '))

    ci = interest.compound_interest(principal, rate, years)
    si = interest.simple_interest(principal, rate, years)
    display.print_result(principal, rate, years, ci, si)

main()
```
