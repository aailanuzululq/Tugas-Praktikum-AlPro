```python
# a. membuat variabel global
skor_nobita = 0

# b. 
def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai
    
# c
def belajar(jam_belajar): # fungsi belajar
    if jam_belajar < 1:
        print('Nobita hanya membuka buku sebentar')
        return 50
    elif jam_belajar < 3: 
        print('Nobita belajar secukupnya')
        return 25
    else:                      
        print('Nobita belajar dengan rajin hari ini')
        return 10


def tidur(jam_tidur): # fungsi jam tidur
    if jam_tidur < 1:
        print('Nobita kurang tidur')
        return -10
    elif jam_tidur <= 2:       
        print('Waktu tidur Nobita seimbang')
        return 20
    else:        
        print('Nobita tidur terlalu lama')
        return -20

def main_bersama_doraemon(jumlah_gadget, waktu_main):  # fungsi main bersama d
    if waktu_main <= 1 and jumlah_gadget <= 2:
        print('Nobita bermain secukupnya')
        return 15
    elif waktu_main <= 2 and jumlah_gadget <= 3:
        print('Nobita masih bisa mengatur waktu dengan baik')
        return 5
    else:
        print('Nobita kebanyakan bermain')
        return -30


def menyontek(): # fungsi menyontek
    print('Nobita ketahuan menyontek')
    return -50

# d. 
def jalankan_aktivitas(pilihan):
    if pilihan == 1:
        jam_belajar = float(input('Berapa jam Nobita belajar hari ini? '))
        update_skor(belajar(jam_belajar))
    elif pilihan == 2:
        jam_tidur = float(input('Berapa jam Nobita tidur siang hari ini? '))
        update_skor(tidur(jam_tidur))
    elif pilihan == 3:
        jumlah_gadget = int(input('Berapa gadget yang digunakan Nobita? '))
        waktu_main = float(input('Berapa jam Nobita bermain gadget? '))
        update_skor(main_bersama_doraemon(jumlah_gadget, waktu_main))
    elif pilihan == 4:
        update_skor(menyontek())
    else:
        print('Pilihan tidak valid, aktivitas hanya ada 4 pilihan')

# e.
def evaluasi_akhir(skor):
    if skor >= 80:
        return 'Siswa Teladan'
    elif skor >= 50:               
        return 'Cukup Rajin'
    else:                          
        return 'Pemalas'


# f. 
def menu_utama():
    lanjut = 'y'
    while lanjut.lower() == 'y':
        print('\n=== PROGRAM PENINGKAT DISIPLIN NOBITA ===')
        print('1. Belajar')
        print('2. Tidur')
        print('3. Main bersama Doraemon')
        print('4. Menyontek saat ujian')
        print()
        pilihan = int(input('Pilih aktivitas (1-4): '))
        jalankan_aktivitas(pilihan)
        print('Skor Nobita saat ini:', skor_nobita)
        lanjut = input('Tambah aktivitas lain? (y/n): ')

    print('\n=== HASIL AKHIR HARI INI ===')
    print('Total skor kedisiplinan:', skor_nobita)
    print('Evaluasi:', evaluasi_akhir(skor_nobita))

# Jalankan program
menu_utama()
```
