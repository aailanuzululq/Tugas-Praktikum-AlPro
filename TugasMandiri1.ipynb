Praktikum: Toko Dorayaki Doraemon

```python
# 1 Pemilihan nama toko yang terdiri atas nama makanan, nama karakter dan kata sifat
nama_makanan = 'Dorayaki' # Membuat variabel nama makanan
nama_karakter = 'Dorami' # Membuat variabel nama karakter
kata_sifat = 'quiet' # Membuat variabel kata sifat
# 2
nama_toko = nama_makanan + nama_karakter + kata_sifat # operasi penggabungan ketiga variabel string 
print('Nama toko dorayaki resmi Doraemon adalah', nama_toko)  # Gunanakan print() untuk hasil output nama toko
```

```python
# 4 Perhitungan total modal awal
Doraemon = 555000  # Variabel modal 1
Nobita = 333000  # Variabel modal 2
Shizuka = 222000  # Variabel modal 3
modal_total = Doraemon + Nobita + Shizuka # operasi untuk perhitungan modal
print('Total modal yang terkumpul untuk membuka toko dorayaki Doraemon adalah', modal_total) # output total modal
```

```python
# 5 Menghitung total pendapatan
jumlah_dorayaki = 350 # Variabel jumlah dorayaki 
harga_dorayaki = 5000 # Variabel harga dorayaki
total_pendapatan = jumlah_dorayaki * harga_dorayaki # operasi penghitungan total pendapatan
print('Total pendapatan Toko DorayakiDoramiquiet dalam bulan pertama adalah', total_pendapatan) # print() untuk hasil output total pendapatan
```

```python
# 6 Menghitung keuntungan bulan pertama
total_pendapatan =  1750000 # variabel total pendapatan
modal_awal = 1100000 # variabel modal awal
keuntungan = total_pendapatan - modal_awal # operasi perhitungan keuntungan
print('Keuntungan toko DorayakiDoramiquiet pada bulan pertama adalah', keuntungan) # untuk hasil output berupa kalimat + total keuntungan
```
